## Setup

This cell checks the environment and creates portable output directories.

```python
# Notebook configuration. Run this cell before the environment check.
import json, os
AEC_EXECUTE = False  # Set True to run training; False only prints the plan.
AEC_GPU_IDS = 'default'  # Use all visible GPUs, or specify [0, 2] or '0,2'.
AEC_MAX_PARALLEL_PER_GPU = 'default'  # Floor(total VRAM / 6 GiB) per GPU; alternatively use 3 or [3, 5].

os.environ['AEC_EXECUTE'] = '1' if AEC_EXECUTE else '0'
os.environ['AEC_GPU_IDS'] = AEC_GPU_IDS if isinstance(AEC_GPU_IDS, str) else json.dumps(AEC_GPU_IDS)
os.environ['AEC_MAX_PARALLEL_PER_GPU'] = AEC_MAX_PARALLEL_PER_GPU if isinstance(AEC_MAX_PARALLEL_PER_GPU, str) else json.dumps(AEC_MAX_PARALLEL_PER_GPU)
print({'AEC_EXECUTE': os.environ['AEC_EXECUTE'], 'AEC_GPU_IDS': os.environ['AEC_GPU_IDS'], 'AEC_MAX_PARALLEL_PER_GPU': os.environ['AEC_MAX_PARALLEL_PER_GPU']})
```

```python
from pathlib import Path
import os, sys
REPO_ROOT = Path.cwd().resolve()
while REPO_ROOT != REPO_ROOT.parent and not (REPO_ROOT / 'scripts' / 'aec').is_dir():
    REPO_ROOT = REPO_ROOT.parent
sys.path.insert(0, str(REPO_ROOT))
from scripts.aec.environment import check_environment
from scripts.aec.paths import ensure_runtime_dirs
ensure_runtime_dirs()
execute = os.environ.get('AEC_EXECUTE', '0') == '1'
ENV = check_environment(require_gpu=execute)
print('Repository:', REPO_ROOT)
```
